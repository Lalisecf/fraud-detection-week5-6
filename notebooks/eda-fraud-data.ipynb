5. Geolocation Integration

```python
#Convert IP Address
fraud_df['ip_address'] = (
    fraud_df['ip_address']
    .astype(float)
    .astype(int)
)
ip_df['lower_bound_ip_address'] = (
    ip_df['lower_bound_ip_address']
    .astype('int64')
)

ip_df['upper_bound_ip_address'] = (
    ip_df['upper_bound_ip_address']
    .astype('int64')
)
```

```python
#Sort Data
fraud_df = fraud_df.sort_values(
    'ip_address'
)

ip_df = ip_df.sort_values(
    'lower_bound_ip_address'
)
```

```python
#Merge Using Range Lookup
country_df = pd.merge_asof(
    fraud_df,
    ip_df,
    left_on='ip_address',
    right_on='lower_bound_ip_address',
    direction='backward'
)
```

```python
#Filter valid ranges:
country_df = country_df[
    country_df['ip_address']
    <= country_df['upper_bound_ip_address']
]
```

```python
#Fraud by Country
country_fraud = (
    country_df
    .groupby('country')['class']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

country_fraud.plot(kind='bar')
```

```python
country_counts = country_df['country'].value_counts().head(10)

country_counts.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top Countries by Transaction Count')
plt.ylabel('Number of Transactions')
plt.show()
```

## Feature Engineering

Feature engineering was performed to create additional variables that capture user behavior, transaction timing, and device usage patterns. These engineered features are expected to improve the model's ability to distinguish between legitimate and fraudulent transactions.

### Engineered Features

#### 1. Time Since Signup

**Feature:** `time_since_signup`

This feature measures the time elapsed between a user's account registration and a purchase transaction.

**Formula:**

Time Since Signup = Purchase Time − Signup Time

The values were converted into seconds for numerical analysis.

**Purpose:**

Fraudsters often create accounts and make purchases shortly afterward. Therefore, a short time interval between signup and purchase may indicate suspicious behavior.

---

#### 2. Hour of Day

**Feature:** `hour_of_day`

This feature extracts the hour (0–23) from the transaction timestamp.

**Purpose:**

Fraudulent transactions may occur more frequently during unusual hours when monitoring activities are reduced. Capturing transaction timing helps identify such temporal patterns.

---

#### 3. Day of Week

**Feature:** `day_of_week`

This feature extracts the day of the week from the transaction timestamp.

**Encoding:**

- 0 = Monday
- 1 = Tuesday
- 2 = Wednesday
- 3 = Thursday
- 4 = Friday
- 5 = Saturday
- 6 = Sunday

**Purpose:**

Customer purchasing behavior often varies throughout the week. Fraudulent activity may also exhibit weekly patterns.

---

#### 4. Transaction Count

**Feature:** `transaction_count`

This feature counts the total number of transactions associated with each user.

**Purpose:**

Users performing an unusually large number of transactions may exhibit suspicious behavior. Transaction frequency is a commonly used fraud detection signal.

---

#### 5. Device Frequency

**Feature:** `device_frequency`

This feature counts how many transactions were performed using the same device.

**Purpose:**

Fraudsters may repeatedly use the same device across multiple accounts or transactions. High device usage frequency can therefore provide valuable fraud indicators.

---

---

## Planned Resampling Strategy

For Task 2, SMOTE (Synthetic Minority Over-sampling Technique) will be applied to the training data only.

SMOTE was selected because:

- It generates synthetic examples of the minority class.
- It increases representation of fraudulent transactions without removing legitimate observations.
- It often improves Recall and F1-Score in fraud detection problems.
- It is widely adopted for highly imbalanced classification tasks.

To avoid data leakage, SMOTE will be applied only after the train-test split and only on the training dataset.

---

## Evaluation Metrics

Because of the severe class imbalance, model performance will be evaluated using:

- Precision
- Recall
- F1-Score
- Precision-Recall AUC (AUC-PR)
- Confusion Matrix

These metrics provide a more reliable assessment of fraud detection performance than overall accuracy.

---

## Conclusion

Both datasets exhibit class imbalance, although the severity differs substantially.

The Fraud_Data dataset contains approximately 9.36% fraudulent transactions, while the Credit Card dataset contains only 0.17% fraudulent transactions. The extreme imbalance in the Credit Card dataset makes fraud detection particularly challenging and reinforces the need for resampling techniques and imbalance-aware evaluation metrics.

SMOTE will be applied during Task 2 after the train-test split to improve the model's ability to identify fraudulent transactions while preventing data leakage.
